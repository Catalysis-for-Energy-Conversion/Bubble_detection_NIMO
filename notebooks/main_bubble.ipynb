```python
import nimo
import os

#目的関数の数を指定します．
ObjectivesNum = 2

#一度のサイクルで人工知能が提案する条件数を指定します．
ProposalsNum = 1

#実験条件をリスト化したファイルを指定
candidates_file = "./candidates.csv"

#人工知能が提案する条件を記載するファイルを指定
proposals_file = "./proposals.csv"

#出力ファイルを格納するフォルダを指定
output_folder = "./EXPOutput"

#計測するフレームを指定
#target_flame = [50, 60, 70]
target_flame = [20 + i*2 for i in range(31)]


files = os.listdir(output_folder)
if '.DS_Store' in files: files.remove('.DS_Store')
if '.ipynb_checkpoints' in files: files.remove('.ipynb_checkpoints')

files_list = [int(f) for f in files]

target_folder = str(max(files_list))

print("name of target folder")
print(output_folder + "/" + target_folder)

import analysis_output_bubble
analysis_output_bubble.ORIGINAL(input_file = proposals_file,
                                  output_file = candidates_file,
                                  num_objectives = ObjectivesNum,
                                  output_folder = output_folder + "/" + target_folder,
                                  target_flame = target_flame).perform()

#人工知能の実行
nimo.selection(method = "BLOX",
               input_file = candidates_file,
               output_file = proposals_file,
               num_objectives = ObjectivesNum,
               num_proposals = ProposalsNum)
```
